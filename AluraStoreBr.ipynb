#1. Análise do faturamento


```python
faturamento = loja['Preço'].sum()
print(f'O faturamento total é de {faturamento}')
```

```python
faturamento = loja2['Preço'].sum()
print(f'O faturamento total é de {faturamento}')
```

```python

# Dados de faturamento
faturamento_lojas = {
    'Loja 1': loja['Preço'].sum(),
    'Loja 2': loja2['Preço'].sum(),
    'Loja 3': loja3['Preço'].sum(),
    'Loja 4': loja4['Preço'].sum()
}

lojas = list(faturamento_lojas.keys())
valores = list(faturamento_lojas.values())

#ajuste do gráfico
plt.figure(figsize=(10, 6))
plt.barh(lojas, valores)
plt.xlabel('Faturamento')
plt.ylabel('Lojas')
plt.title('Faturamento por Loja')


# Adiciona os valores de faturamento em cima de cada barra
for i, v in enumerate(valores):
    plt.text(v + 10, i, str(round(v, 2)), color='black', va='center')

plt.show()
```

# 2. Vendas por Categoria


```python
categorias = loja['Categoria do Produto']
categorias.value_counts()
```

```python
categorias = loja2['Categoria do Produto']
categorias.value_counts()
```

```python
categorias = loja3['Categoria do Produto']
categorias.value_counts()
```

```python
categorias = loja4['Categoria do Produto']
categorias.value_counts()
```

```python
#função para plotar o gráfico categoria de produto
def plot_category_sales(df, loja_num):
    category_counts = df.groupby('Categoria do Produto').size()
    return category_counts

loja1_categories = plot_category_sales(loja, 1)
loja2_categories = plot_category_sales(loja2, 2)
loja3_categories = plot_category_sales(loja3, 3)
loja4_categories = plot_category_sales(loja4, 4)

#concatenando as categorias de cada loja
all_categories = pd.concat([loja1_categories, loja2_categories, loja3_categories, loja4_categories], axis=1)
all_categories.columns = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

all_categories.plot(kind='bar', figsize=(12, 6), color=['red', 'green', 'blue', 'orange'])
plt.title('Vendas por Categoria de Produto em Todas as Lojas')
plt.xlabel('Categoria do Produto')
plt.ylabel('Número de Vendas')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Lojas')
plt.tight_layout()
plt.show()
```

# 4. Produtos Mais e Menos Vendidos

```python
#função para selecionar o três produtos mais vendidos de cada loja
def top3_products(df, loja_num):
    top_products = df.groupby('Produto').size().nlargest(3)
    top_products = top_products.reset_index()
    top_products = top_products.rename(columns={0: 'Qtd'})
    top_products['Loja'] = f'{loja_num}'
    return top_products

top3_loja1 = top3_products(loja, 1)
top3_loja2 = top3_products(loja2, 2)
top3_loja3 = top3_products(loja3, 3)
top3_loja4 = top3_products(loja4, 4)

#concatenar lado a lado
top3_all_stores = pd.concat([top3_loja1, top3_loja2, top3_loja3, top3_loja4], axis=1)
top3_all_stores
```

```python
#função para selecionar o três produtos menos vendidos de cada loja
def top3_products(df, loja_num):
    top_products = df.groupby('Produto').size().tail(3)
    top_products = top_products.reset_index()
    top_products = top_products.rename(columns={0: 'Qtd'})
    top_products['Loja'] = f'{loja_num}'
    return top_products

top3_loja1 = top3_products(loja, 1)
top3_loja2 = top3_products(loja2, 2)
top3_loja3 = top3_products(loja3, 3)
top3_loja4 = top3_products(loja4, 4)

#concatenar lado a lado
top3_all_stores = pd.concat([top3_loja1, top3_loja2, top3_loja3, top3_loja4], axis=1)
top3_all_stores
```

# 5. Frete Médio por Loja


```python
#extraindo a média do frete de cada loja e regando o gráfico respectivamente
frete_medio_lojas = {
    'Loja 1': loja['Frete'].mean(),
    'Loja 2': loja2['Frete'].mean(),
    'Loja 3': loja3['Frete'].mean(),
    'Loja 4': loja4['Frete'].mean()
}

lojas = list(frete_medio_lojas.keys())
valores = list(frete_medio_lojas.values())

plt.figure(figsize=(10, 6))
plt.bar(lojas, valores)
plt.xlabel('Lojas')
plt.ylabel('Frete Médio')
plt.title('Frete Médio por Loja')

# Adiciona os valores de frete médio em cima de cada barra
for i, v in enumerate(valores):
    plt.text(i, v + 0.1, str(round(v, 2)), ha='center', va='bottom')

plt.show()
```

O fato de ter um  baixo faturamento impacta muito na decisão que o Sr. João terá que tomar, as causas possíveis que levam a ter um baixo desempenho nesse quesito:

- Produtos com menor valor agregado

Mesmo com um alto volume de vendas, os itens vendidos podem ser mais baratos em relação às outras.

- Localização ou público-alvo

A localização pode atrair menos clientes ou um público pode gastar menos.

- Método de pagamento

Com a opção de crédito se não houver limites parcelamentos, os clientes podem parcelar muito e a consequência disso é o juros da conversão desse crédito.
